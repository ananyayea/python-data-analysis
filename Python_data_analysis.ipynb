```python
df = df.drop(['Row ID', 'Order ID',  'Customer ID', 'Postal Code'], axis = 1)
```

```python
df
```

```python
cat_profit = df.groupby('Category')['Profit'].sum()
```

```python
cat_profit.plot(kind = 'bar')
plt.ylabel('Profit')
```

```python
df.head()
```

```python
reg_sales = df.groupby('Region')['Sales'].sum()
reg_sales.plot(kind = 'bar')
plt.xlabel('Region')
plt.ylabel('Total Sales')
#the hypothesis not supported as the central region has the highest sales
```

```python
df.head()
```

```python
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    format='mixed',
    dayfirst=True
)
```

```python
df['Month'] = df['Order Date'].dt.month
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales.plot(kind = 'line')
plt.xlabel('Months')
plt.ylabel('Total sales')
#conclusion: The hypothesis is supported since the sales are higher dusring certain months
```

```python
total_shipping_orders = df.groupby('Ship Mode').size()
returned_orders = df[df['Profit'] < 0].groupby('Ship Mode').size()
percent = (returned_orders/total_shipping_orders) * 100
```

```python
percent
percent.plot(kind = 'bar')
plt.ylabel('Return Mode')
```
